Aujourd'hui, nous apprenons à regrouper et agréger des données, une étape clé dans la préparation des données pour l'Intelligence Artificielle.

## 1. GroupBy (Regroupement)
Le `.groupby()` permet de regrouper des lignes identiques et d'appliquer une fonction mathématique (comme la moyenne `.mean()` ou la somme `.sum()`).

```python
import pandas as pd

df = pd.DataFrame({
    'Service': ['RH', 'IT', 'RH', 'IT', 'IT'],
    'Salaire': [3000, 4000, 3200, 4500, 4200]
})

# Quelle est la moyenne des salaires par service ?
moyenne_service = df.groupby('Service')['Salaire'].mean()
moyenne_service
```

## 2. Gestion des valeurs manquantes
Dans la vraie vie, les données sont souvent incomplètes. Pandas représente ces vides par `NaN` (Not a Number). On peut les remplir avec `.fillna()` ou les supprimer avec `.dropna()`.

```python
df_manquant = pd.DataFrame({
    'A': [1, 2, None],
    'B': [4, None, 6]
})

print("DataFrame avec manquants :")
print(df_manquant)

print("\nDataFrame rempli avec des 0 :")
print(df_manquant.fillna(0))
```

## 🏆 Défi du Jour 11

1. Créez un DataFrame avec des étudiants, leurs `Matiere` (Maths, Physique, Maths) et leurs `Note` (15, 12, 18).
2. Utilisez `groupby` pour calculer la **moyenne** des notes par matière.
3. Affichez le résultat.

```python
import pandas as pd

# 1. Création du DataFrame des étudiants
df_etudiants = pd.DataFrame({
    'Etudiant': ['Awa', 'Moussa', 'Fatou'],
    'Matiere': ['Maths', 'Physique', 'Maths'],
    'Note': [15, 12, 18]
})
print(df_etudiants)

# 2. Regroupement par Matiere et calcul de la moyenne des notes
moyenne_matiere = df_etudiants.groupby('Matiere')['Note'].mean()

# 3. Affichage du résultat
print("\nMoyenne des notes par matière :")
print(moyenne_matiere)
```
